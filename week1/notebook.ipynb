```python
from openai import OpenAI
from toyaikit.llm import OpenAIClient
from toyaikit.chat import IPythonChatInterface
from toyaikit.chat.runners import OpenAIResponsesRunner
```

```python
client = OpenAI()
```

```python
messages = [
        {
            "role": "system",
            "content": "You're an assistant that makes the user laugh. Include emojis in every response."
        },
        {
            "role": "user",
            "content": "tell me a joke about Dan Le Batard"
        }
    ]

response = client.responses.create(
    model="gpt-4o-mini",
    input=messages
)

print(response.output_text)
```

```python
# print output tokens as soon as they are generated
messages = [
        {
            "role": "system",
            "content": """
                You're an assistant that creates bedtime stories based on user input. Include emojis in every response.
                Ask the user follow up questions and make sure the stories are presonalized.
            """
        },
        {
            "role": "user",
            "content": "unicorn"
        }
    ]

stream = client.responses.create(
    model="gpt-4o-mini",
    input=messages,
    stream=True
)

for event in stream:
    if hasattr(event, 'delta'):
        print(event.delta, end='')
```

```python
system_prompt = """
    you're an assistant that creates bedtime stories. Include emojis in the response.
    Ask the user follow up questions and make the stories are personalized.
""".strip()

llm_client = OpenAIClient(
    model="gpt-4o-mini",
    client=client
)
```

```python
runner = OpenAIResponsesRunner(
    tools=None,
    developer_prompt=system_prompt,
    chat_interface=IPythonChatInterface(),
    llm_client=llm_client
)
```

```python
runner.run();
```
