```python
from openai import OpenAI
```

```python
client = OpenAI()
```

```python
messages = [
        {
            "role": "system",
            "content": "You're an assistant that makes the user laugh. Include emojis in every response."
        },
        {
            "role": "user",
            "content": "tell me a joke about Dan Le Batard"
        }
    ]

response = client.responses.create(
    model="gpt-5",
    input=messages
)

print(response.output_text)
```

```python
# streaming
messages = [
        {
            "role": "system",
            "content": "You're an assistant that creates bedtime stories based on user input. Include emojis in every response."
        },
        {
            "role": "user",
            "content": "unicorn"
        }
    ]

stream = client.responses.create(
    model="gpt-4o-mini",
    input=messages,
    stream=True
)

for event in stream:
    if hasattr(event, 'delta'):
        print(event.delta, end='')
```
