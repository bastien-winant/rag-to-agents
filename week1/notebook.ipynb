```python
from openai import OpenAI
```

```python
openai_client = OpenAI()
```

```python
# get a single prompt response from the LLM
messages = [
    {
        # instruction for the agent
        "role": "system",
        "content": "you're an assistant that makes the user laugh. Include emojis in the response."
    },
    {
        "role": "user",
        "content": "tell me a joke about Obama"
    }
]

response = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages,
)

print(response.output_text)
```

```python
# print response tokens as soon as they are available
stream = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages,
    stream=True
)

for event in stream:
    if hasattr(event, 'delta'):
        print(event.delta, end='')
```

```python
instructions = """
    You're an assistant that creates bedtime stories given a subject. Ask the user follow-up questions and make sure the stories are personalized.
""".strip()
# get a single prompt response from the LLM
messages = [
    {
        # instruction for the agent
        "role": "system",
        "content": instructions
    },
    {
        "role": "user",
        "content": "unicorn"
    }
]

response = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages,
)

print(response.output_text)
```

> LLM are inherently stateless: every prompt lives in a vacuum. To enable conversation, the LLM's previous response needs to be passed back in the next prompt.

```python
messages.append(response.output[0])
messages.append({"role" : "user", "content": "Ashley is a unicorn that lives by the beach and starts an adventure when meeting a beaver"})
```

```python
response = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages,
)

print(response.output_text)
```

```python
from toyaikit.llm import OpenAIClient
from toyaikit.chat import IPythonChatInterface
from toyaikit.chat.runners import OpenAIResponsesRunner
```

```python
system_prompt = """
you're an assistant that can make jokes. Always find out the name of
the person to make the jokes personalized. Once you know the name,
make the joke about them.
""".strip()

llm_client = OpenAIClient(
    model='gpt-4o-mini',
    client=openai_client
)

runner = OpenAIResponsesRunner(
    tools=None,
    developer_prompt=system_prompt,
    chat_interface=IPythonChatInterface(),
    llm_client=llm_client
)

runner.run();
```
