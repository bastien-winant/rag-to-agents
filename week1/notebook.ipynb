```python
from openai import OpenAI
```

```python
client = OpenAI()
```

### Basic LLM prompt

```python
# simple prompt-response
response = client.responses.create(
    model="gpt-4o-mini",
    input="Write a short bedtime story about a unicorn."
)

print(response.output_text)
```

```python
# store messages in a separate array
messages = [
    {
        "role": "system",
        "content": "you're an assistent that makes the user laugh. Include emojis in the response."
    },
    {
        "role": "user",
        "content": "tell me a joke about Alexey"
            
    }
]
```

```python
response = client.responses.create(
    model="gpt-5",
    input=messages
)

print(response.output_text)
```

```python
# inspect response object
response.output[1].content[0].text
```

### Output streaming

```python
messages = [
    {
        "role": "system",
        "content": "You're an assistant that creates bedtime stories based on user-supplied theme. Inlcude emojis in the response"
    },
    {
        "role": "user",
        "content": "unicorn"
    }
]
```

```python
# set up reponses as stream
stream = client.responses.create(
    model="gpt-4o-mini",
    input=messages,
    stream=True
)
```

```python
# print the output as soon as it is generated
for event in stream:
    if hasattr(event, 'delta'):
        print(event.delta, end='')
```

### Context accumulation

```python
system_prompt = """
You're an assistant that can make jokes. Always find out the name of
the person to make the jokes personalized. Once you know the name,
make the joke about them.
""".strip()

messages = [
    {"role": "developer", "content": system_prompt},
    {"role": "user", "content": "tell me a joke"}
]

response = client.responses.create(
    model='gpt-4o-mini',
    input=messages
)
```

```python
# feed the LLM response back into the context
messages.append(response.output[0])
print(messages)

# add the next user prompt
messages.append({"role": "user", "content": "Alexey"})
```

```python
response = client.responses.create(
    model='gpt-4o-mini',
    input=messages
)

print(response.output_text)
```

### ToyAIKit

```python
from toyaikit.llm import OpenAIClient
from toyaikit.chat import IPythonChatInterface
from toyaikit.chat.runners import OpenAIResponsesRunner

system_prompt = """
    you're an assistant that can make jokes. Always find out the name of
    the person to make the jokes personalized. Once you know the name,
    make the joke about them.
""".strip()

llm_client = OpenAIClient(
    model='gpt-4o-mini',
    client=client
)

runner = OpenAIResponsesRunner(
    tools=None,
    developer_prompt=system_prompt,
    chat_interface=IPythonChatInterface(),
    llm_client=llm_client
)

runner.run();
```
