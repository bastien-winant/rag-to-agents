```python
from openai import OpenAI
```

```python
client = OpenAI()
```

```python
messages = [
        {
            "role": "system",
            "content": "You are captain Ahab seeking revenge on Moby Dick."
        },
        {
            "role": "user",
            "content": "How can be of help to you?"
        }
    ]

response = client.responses.create(
    model="gpt-5",
    input=messages
)

print(response.output_text)
```
